## 2. The question: decision, action, cost of a wrong call

**Research question:** Which pages should an editor review first for a potential content refresh, using observable search and content signals available before the review decision?

**Decision:** prioritize a limited set of pages for refresh review.

**Who acts:** an editor or SEO/content team member reviews the ranked queue and decides whether to update a page.

**Action:** inspect the highest-priority pages, then refresh content only when the evidence supports it.

**Cost of a wrong call:** a false positive wastes editorial time and may lead to an unnecessary update; a false negative can leave a genuinely declining page unreviewed. Because review capacity is limited, ranking quality at the top of the queue matters.

**Why data/ML helps:** a simple rule such as stale × visible is a useful baseline, but multiple signals can interact. ML is useful only if it consistently improves the ranking over a transparent baseline on an honest holdout.


```python
# Code cell: state the decision and unit of analysis.
unit_of_analysis = 'content item / page'
decision = 'which pages should be reviewed first for refresh'
print('Unit:', unit_of_analysis)
print('Decision:', decision)
```

## 3. Quick look at the data (2-3 real numbers)

The starter CSV is the evidence base for this provisional choice. The code below calculates the numbers live rather than hardcoding them.


```python
import pandas as pd

df = pd.read_csv('data/raw/content_refresh_anonymized.csv')
df['is_declining_label'] = df['trend_direction'].str.lower().eq('down').astype(int)

stale_visible = (df['days_since_last_update'] >= 180) & (df['impressions_90d'] >= 500)
print(f'Rows in starter dataset: {len(df):,}')
print(f'Declining rate: {df["is_declining_label"].mean():.3f}')
print(f'Stale + visible pages (>=180 days since update and >=500 impressions): {stale_visible.sum():,}')
```

These numbers make the lane worth investigating: there is a substantial page-level sample, the starter proxy label has a measurable declining group, and a concrete subset of stale but still visible pages exists for review prioritization. These are evidence for investigation, not proof that refreshing a page causes recovery.
